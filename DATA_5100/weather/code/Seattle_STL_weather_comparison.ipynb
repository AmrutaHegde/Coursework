## Join data frames keeping `DATE` and `PRCP` columns

```python
df_merged = stl_data[['DATE', 'PRCP']].merge(seattle_data[['DATE', 'PRCP']], on='DATE', how='outer', suffixes=('_STL', '_SEA'))
```

```python
df_merged.head()
```

**What type of join should we do?**

##### Use a join to keep all dates present

Each DataFrame had a column named `PRCP`, so the default is to add suffixes _x and _y to differentiate the columns. Normally, I would rename the columns at this point to something more informative. However, I know that I want to convert the DataFrame to a tidy format and I will modify the names later.

### Create a tidy data frame with columns for city and precipitation

```python
df = pd.melt(df_merged, id_vars=['DATE'], value_vars=['PRCP_STL', 'PRCP_SEA'], var_name='City', value_name='Precipitation')
```

### Impute missing values

We will replace missing values with the mean across years of values on that day.


**Design an algorithm for replacing missing values with the mean across years of values on that day.**

```python
df['day_of_year'] = pd.DatetimeIndex(df['Date']).dayofyear
```

```python
df.head()
```

```python


# mean_day_precipitation = df.loc[df['City']=='SEA', 'Precipitation', 'day_of_year'].groupby(df['day_of_year']).mean()

mean_day_precipitation=df.loc[
    df['City']=='SEA',
    ['Precipitation','day_of_year']
].groupby(
    'day_of_year'
).mean()
mean_day_precipitation
```

```python
df.head()
```

```python
mean_day_precipitation=df.loc[
    df['City']=='STL',
    ['Precipitation','day_of_year']
].groupby(
    'day_of_year'
).mean()
```

```python
df
```

```python
plt.figure(figsize=(20,5))

sns.lineplot(data=mean_day_precipitation, x='day_of_year', y='Precipitation')
plt.xlabel('Day of Year', fontsize=15)
plt.ylabel('Mean Precipitation (inches)', fontsize=15)
plt.tick_params(labelsize=15)   
plt.show()
```

```python
indices = np.where(df['Precipitation'].isna() == True)[0]
indices
```

```python
for index in indices:
    df.loc[index,'Precipitation'] = mean_day_precipitation.loc[df.loc[index, 'day_of_year']].values[0]
```

Plot the cleaned csv to visualise the precipitations throughout the year

```python

plt.figure(figsize=(20,5))
df = pd.read_csv(DATA_DIR / "clean_seattle_stlouis_weather.csv")

df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year

yearly_rain = (
    df.groupby(["Year", "City"])["Precipitation"]
      .sum()
      .reset_index()
)

for city in yearly_rain["City"].unique():
    city_data = yearly_rain[yearly_rain["City"] == city]
    plt.plot(
        city_data["Year"],
        city_data["Precipitation"],
        marker="o",
        label=city
    )

plt.title("Yearly Precipitation by City (2018–2022)")
plt.xlabel("Year")
plt.ylabel("Total Precipitation")
plt.xticks(range(2018, 2023))
plt.legend()
plt.tight_layout()
plt.show()
```

Plot the cleaned csv to visualise the precipitations throughout the year Monthly


```python
df["Month"] = df["Date"].dt.to_period("M")
plt.figure(figsize=(20,5))
monthly_rain = (
    df.groupby(["Month", "City"])["Precipitation"]
      .sum()
      .reset_index()
)

monthly_rain["Month"] = monthly_rain["Month"].dt.to_timestamp()

for city in monthly_rain["City"].unique():
    city_data = monthly_rain[monthly_rain["City"] == city]

    plt.plot(
        city_data["Month"],
        city_data["Precipitation"],
        label=city
    )

plt.title("Monthly Precipitation by City (2018–2022)")
plt.xlabel("Date")
plt.ylabel("Total Monthly Precipitation")
plt.legend()
plt.tight_layout()
plt.show()
```

### Conclusion

From 2018 to 2022, monthly precipitation varied in both Seattle and St. Louis, and neither station recorded more in every month. However, the yearly totals indicate that the St. Louis station recorded more precipitation overall during this period. This comparison measures total precipitation, not the number of rainy days, and represents these stations rather than each entire city.
